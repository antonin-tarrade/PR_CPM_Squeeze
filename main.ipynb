# OBG TO OBJA TRANSFORMATION

## Loading model

```python
# notebook.ipynb
from MeshObjects import *
from utils import show_obj
import trimesh

obj_path = "./models/bunny.obj"


objref = trimesh.load(obj_path)
objA = AObject3D(objref, "Object A") # M_max
show_obj(objref, "Original Object (M_max)")

objA.compress(0.1, 5)
show_obj(objA.get_last_lod(), "Compressed Object (M_min)", point="both")

M_max_prime = objA.decompress() # M_max'
show_obj(M_max_prime, "Reconstructed Object (M_max')", point="both")
```
