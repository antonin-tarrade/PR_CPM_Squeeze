# OBG TO OBJA TRANSFORMATION

## Loading model

```python
# notebook.ipynb
from MeshObjects import *

# Parramètres
obj_name = "icosphere"
obj_path = f"./models/{obj_name}.obj"
color = "#00FF04"

# Model Ref = M_max = M_n
with open(obj_path, 'r') as f:
    objref = Object3D(f, "Reference")
objref.Show(color)

objA = AObject3D(objref, "Object A") # M_max

# Model compression to M_0
objA.compress(0.1, 20)
objA.show_last_lod(color)
objA.get_last_lod().export_as_obj(obj_name)

# Model decompression to M_max'
M_max_prime = objA.decompress() # M_max'
M_max_prime.Show(color, "Reconstructed Object (M_max')")
M_max_prime.export_as_obj(obj_name)

```
