# OBG TO OBJA TRANSFORMATION

## Loading model

```python
from MeshObjects import * 
from obja.obja import Output 
import os 

# Parramètres 
obj_name = "cow" 
color = "#EA00FF"
compression_ratio = 0.4
nombre_steps = 10

# Paths
model_path = f"models/{obj_name}.obj"
export_dir = f"./export/{obj_name}"
os.makedirs(export_dir, exist_ok=True) 
obj_path = f"{export_dir}/{obj_name}.obj" 
obja_path = f"{export_dir}/{obj_name}.obja" 

# Model Ref = M_max = M_n 
with open(model_path, 'r') as f: 
    objref = Object3D(f, "Reference") 
objref.Show(color) 
LODs = AObject3D(objref, "Object A") # M_max 

# Model compression to M_0 
LODs.compress(compression_ratio, nombre_steps)
LODs.show_last_lod(color) 
LODs.get_last_lod().export_as_obj(obj_name)

# Model decompression to M_max'
output = Output(open(obja_path, "w"))
M_max_prime = LODs.decompress(out=output) 
M_max_prime.Show(color, "Reconstructed Object (M_max')") 
M_max_prime.export_as_obj(obj_name)
```
