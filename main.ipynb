# OBG TO OBJA TRANSFORMATION

## Loading model

```python
# notebook.ipynb
from Decompression import decompress
from MeshObjects import *

obj_path = "./models/bunny.obj"

with open(obj_path, 'r') as f:
    objref = Object3D(f, "Reference")
    objref.Show()

objA = AObject3D(objref, "Object A")
objA.compress(0.1, 9)
M0 = objA.get_last_lod()
collapse_info = objA.collapse_info[-1]
adjacency = objA.adjacency_ref[-1]
restored_obj = decompress(M0, adjacency ,collapse_info)


objA.show_last_lod () # M0
objA.model_lods[-2].Show("Compressed Object (LOD)") # M1 compressed
restored_obj.Show("Reconstructed Object (Decompressed)") # M1 reconstructed



```
